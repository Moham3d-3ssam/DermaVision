```python
# Import Edited Dataset

import pickle

with open("../data/features/dataset.pkl", "rb") as f:
  dataset = pickle.load(f)
```

```python
# Get New Columns

import pandas as pd

group_1 = dataset.copy()
group_2 = dataset.copy()
group_3 = dataset.copy()

def change_img_path(data, path):
  for i in range(len(data)):
    data["img_id"][i] = f"../data/imgs/{path}/{data['img_id'][i]}"

change_img_path(dataset, "resized_imgs")
change_img_path(group_1, "flip_noise_color_imgs")
change_img_path(group_2, "flip_blur_imgs")
change_img_path(group_3, "rotate_zoom_imgs")
```

```python
# Combine All Datasets + Shuffle

allDataset = pd.concat([dataset, group_1, group_2, group_3], ignore_index = True)
allDataset = allDataset.sample(frac = 1).reset_index(drop = True)

print(allDataset.shape)
print(allDataset.keys())
```

```python
# Split Data Into (CNN + ANN), Final Classifier

total_rows = len(allDataset)
two_third_rows = int(0.6 * total_rows)

train_dataset = allDataset.iloc[:two_third_rows].reset_index(drop = True)
ann_train = train_dataset.drop(columns = ["background_father", "background_mother", "img_id"])
cnn_train = train_dataset[["img_id", "diagnostic"]]

final_dataset = allDataset.iloc[two_third_rows:].reset_index(drop = True)
ann_final = final_dataset.drop(columns = ["background_father", "background_mother", "img_id"])
cnn_final = final_dataset[["img_id", "diagnostic"]]
```

```python
# Export ALL Dataset

import pickle

with open("../data/features/allDataset.pkl", "wb") as f:
  pickle.dump(allDataset, f)

with open("../data/features/ann_train.pkl", "wb") as f:
  pickle.dump(ann_train, f)

with open("../data/features/cnn_train.pkl", "wb") as f:
  pickle.dump(cnn_train, f)

with open("../data/features/ann_final.pkl", "wb") as f:
  pickle.dump(ann_final, f)

with open("../data/features/cnn_final.pkl", "wb") as f:
  pickle.dump(cnn_final, f)
```

```python
# Save All Dataset

allDataset.to_csv("allDataset.csv")
```
