```python
import numpy as np
import random
import os
import cv2
```

```python
# Read Images To Resize

for img in os.listdir("../data/imgs/main_imgs"):
  image_path = os.path.join("../data/imgs/main_imgs", img)
  myImg = cv2.imread(image_path)

  resized = cv2.resize(myImg, (224, 224), interpolation = cv2.INTER_LINEAR)

  cv2.imwrite(f"../data/imgs/resized_imgs/{img}", resized)
```

```python
# Vertical Flip + Gaussian Noise + Color => Create new images

for img in os.listdir("../data/imgs/resized_imgs"):
  image_path = os.path.join("../data/imgs/resized_imgs", img)
  myImg = cv2.imread(image_path)

  flipped = cv2.flip(myImg, 1)
  noise_image = np.clip(flipped + np.random.normal(0, 15, flipped.shape), 0, 255).astype(np.uint8)
  changeColors = cv2.convertScaleAbs(noise_image, alpha = 1.2, beta = 30)
  resized = cv2.resize(changeColors, (224, 224), interpolation = cv2.INTER_LINEAR)

  cv2.imwrite(f"../data/imgs/flip_noise_color_imgs/{img}", resized)
```

```python
# Horizontal Flip + Gaussian Blur => Create new images

for img in os.listdir("../data/imgs/resized_imgs"):
  image_path = os.path.join("../data/imgs/resized_imgs", img)
  myImg = cv2.imread(image_path)

  flipped = cv2.flip(myImg, 0)
  blurred = cv2.GaussianBlur(flipped, (5, 5), 0)
  resized = cv2.resize(blurred, (224, 224), interpolation = cv2.INTER_LINEAR)

  cv2.imwrite(f"../data/imgs/flip_blur_imgs/{img}", resized)
```

```python
# Rotate + Zoom => Create new images

for img in os.listdir("../data/imgs/resized_imgs"):
  image_path = os.path.join("../data/imgs/resized_imgs", img)
  myImg = cv2.imread(image_path)

  angle = random.uniform(-15, 15)
  h, w = myImg.shape[:2]
  M = cv2.getRotationMatrix2D((w // 2, h // 2), angle, 1.0)
  rotated = cv2.warpAffine(myImg, M, (w, h), borderMode = cv2.BORDER_REFLECT101)

  scale = random.uniform(0.8, 1.0)
  new_w, new_h = int(w * scale), int(h * scale)
  x1 = random.randint(0, w - new_h)
  y1 = random.randint(0, h - new_h)
  cropped = rotated[y1:y1+new_h, x1:x1+new_w]

  resized = cv2.resize(cropped, (224, 224), interpolation = cv2.INTER_LINEAR)

  cv2.imwrite(f"../data/imgs/rotate_zoom_imgs/{img}", resized)
```
